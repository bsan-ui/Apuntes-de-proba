# Practica 1 -- Experimentos --

## Experimento Determinista
> Es aquel que da el mismo resultado bajo las mismas condiciones iniciales. si las variables usadas se mantienen, el resultado tambien lo hara.\
Por ejemplo:
* $ 2 + 3 = 5 $
* $ A = π ⋅ r^2$





### Ejemplo simbólico

```python
import sympy

def area(radio):
  return sympy.pi * (radio ** 2)
```

```python
area(10)
```

### Ejemplo numérico

```python
import sympy

def areaNum(radio):
  return 3.1416 * (radio ** 2)
```

```python
areaNum(10)
```

## Experimento aleatorio
>Es aquel que aun repitiendo bajo las mismas condiciones, no brinda el mismo resultado siempre, y no es posible predecir con certeza que se obtendra.

### Ejemplo de lanzamiento de moneda

```python
import random as rnd

monedaLados = ["aguila", "sol"]

def lanzarMoneda():
  return rnd.choice(monedaLados)
```

```python
lanzarMoneda()
```

### Ejemplo de lanzamiento de dado

```python
import random as rnd

dadoLados = [1,2,3,4,5,6]

def lanzarDado():
  return rnd.choice(dadoLados)
```

```python
lanzarDado()
```
